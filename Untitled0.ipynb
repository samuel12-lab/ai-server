```python
import os
import time
import datetime
import re
import subprocess
import torch
import numpy as np
import gradio as gr
from google.colab import drive

# =========================================================
# STEP 1: Mount Google Drive
# =========================================================
# This will prompt you to grant Colab access to your Drive
drive.mount('/content/drive')

# =========================================================
# STEP 2: Setup Cloudflare Tunnel for Mobile Access
# =========================================================
# Close any active Gradio server instances to free port 7860
gr.close_all()
subprocess.run(["pkill", "-f", "cloudflared"])

if not os.path.exists('/usr/local/bin/cloudflared'):
    print("Installing Cloudflare Tunnel...")
    subprocess.run(["wget", "-q", "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64", "-O", "/usr/local/bin/cloudflared"])
    subprocess.run(["chmod", "+x", "/usr/local/bin/cloudflared"])

# Start fresh tunnel on port 7860
tunnel = subprocess.Popen(["cloudflared", "tunnel", "--url", "http://localhost:7860"], stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

time.sleep(4)

public_url = None
for _ in range(20):
    line = tunnel.stderr.readline()
    if not line:
        break
    match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
    if match:
        public_url = match.group(0)
        break

print("\n" + "="*60)
if public_url:
    print(f"🌐 CLOUDFLARE PUBLIC LINK FOR MOBILE:\n👉 {public_url}")
else:
    print("⚠️ Tunnel initialization complete.")
print("="*60 + "\n")

# =========================================================
# STEP 3: Photonic Engine v2.8 (Vectorized Execution)
# =========================================================
SAVE_DIR = '/content/drive/MyDrive/PhotonicEngine_Outputs'
os.makedirs(SAVE_DIR, exist_ok=True)

class PhotonicEngineV2_8:
    def __init__(self, save_dir=SAVE_DIR):
        self.device = "cuda" if torch.cuda.is_available() else "cpu"
        self.save_dir = save_dir

    def run_loss_mismatch_test(self, num_trials=1000, actual_loss=0.99, estimated_loss=0.98):
        A = np.random.uniform(-5, 5, size=(num_trials, 3, 3)).astype(np.float32)
        B = np.random.uniform(-5, 5, size=(num_trials, 3, 3)).astype(np.float32)

        reference_results = A @ B
        U, S, Vh = np.linalg.svd(A)

        # 1. Ideal Baseline
        Sigma_ideal = S[..., None] * np.eye(3)
        ideal_matrices = (U @ Sigma_ideal @ Vh) @ B
        ideal_errors = np.max(np.abs(ideal_matrices - reference_results), axis=(1, 2))

        # 2. Uncorrected Physical Transmission Loss
        Sigma_actual = (S * actual_loss)[..., None] * np.eye(3)
        uncorrected_matrices = np.real((U @ Sigma_actual @ Vh) @ B)
        uncorrected_errors = np.max(np.abs(uncorrected_matrices - reference_results), axis=(1, 2))

        # 3. Inverse Scaling via Mismatched Calibration Factor
        corrected_matrices = uncorrected_matrices / estimated_loss
        mismatch_corrected_errors = np.max(np.abs(corrected_matrices - reference_results), axis=(1, 2))

        # Boundary Verification
        phases = np.arccos(np.clip(np.real(np.diagonal(U, axis1=-2, axis2=-1)), -1.0, 1.0))
        phase_violations = np.any(np.abs(phases) > 2 * np.pi, axis=1)
        sv_violations = np.any(S > 15.0, axis=1)
        total_failures = int(np.sum(phase_violations | sv_violations))

        report = {
            "trials": num_trials,
            "actual_loss": actual_loss,
            "estimated_loss": estimated_loss,
            "mean_ideal_error": float(np.mean(ideal_errors)),
            "mean_uncorrected_error": float(np.mean(uncorrected_errors)),
            "mean_mismatch_error": float(np.mean(mismatch_corrected_errors)),
            "max_mismatch_error": float(np.max(mismatch_corrected_errors)),
            "phase_failures": total_failures
        }

        saved_filepath = self._save_report_to_drive(report)
        report["saved_filepath"] = saved_filepath
        return report

    def _save_report_to_drive(self, report):
        timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S_%f")
        filepath = os.path.join(self.save_dir, f"v2.8_loss_mismatch_{timestamp}.txt")

        log_text = (
            f"--- Photonic Engine v2.8 Loss Estimation Mismatch Report ---\n"
            f"Timestamp: {timestamp}\n"
            f"Trials: {report['trials']}\n"
            f"Actual Transmission: {report['actual_loss']}\n"
            f"Estimated Transmission: {report['estimated_loss']}\n"
            f"Mean Ideal Error: {report['mean_ideal_error']:.8f}\n"
            f"Mean Uncorrected Error: {report['mean_uncorrected_error']:.8f}\n"
            f"Mean Mismatch-Corrected Error: {report['mean_mismatch_error']:.8f}\n"
            f"Maximum Mismatch Error: {report['max_mismatch_error']:.8f}\n"
            f"Phase-Range Failures: {report['phase_failures']}\n"
        )

        with open(filepath, "w") as f:
            f.write(log_text)

        return filepath

engine = PhotonicEngineV2_8()

# =========================================================
# STEP 4: Interface Construction & Launch
# =========================================================
def run_v2_8_ui(actual_loss, estimated_loss):
    res = engine.run_loss_mismatch_test(num_trials=1000, actual_loss=actual_loss, estimated_loss=estimated_loss)

    text_report = (
        f"⚖️ Loss Estimation Mismatch Test Completed (1,000 Trials)\n\n"
        f"1. Mean Ideal Error (Baseline): {res['mean_ideal_error']:.8f}\n"
        f"2. Mean Uncorrected Error: {res['mean_uncorrected_error']:.8f}\n"
        f"3. Mean Mismatch-Corrected Error: {res['mean_mismatch_error']:.8f}\n"
        f"4. Maximum Mismatch Error: {res['max_mismatch_error']:.8f}\n"
        f"5. Phase-Range Failures: {res['phase_failures']}/1,000\n"
    )

    return text_report, res['saved_filepath']

demo = gr.Interface(
    fn=run_v2_8_ui,
    inputs=[
        gr.Slider(0.90, 1.0, value=0.99, label="Actual Physical Transmission Factor"),
        gr.Slider(0.90, 1.0, value=0.98, label="Estimated Calibration Transmission Factor")
    ],
    outputs=[
        gr.Textbox(label="Output Data"),
        gr.File(label="Download Full Report (.txt)")
    ],
    title="Photonic Computing Engine v2.8",
    description="Evaluates calibration estimation error by applying a mismatched transmission factor during inverse-scaling compensation across 1,000 trials."
)

demo.launch(server_name="0.0.0.0", server_port=7860, share=False)
```
